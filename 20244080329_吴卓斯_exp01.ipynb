### 实验2 函数、条件、循环和递归

任务1：写 score_level(score)，按分数返回对应等级；分数不在 0~100 之间就抛 ValueError。


```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("成绩必须在 0~100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in [95, 88, 70, 60, 59, 0]:
    print(score, "->", score_level(score))

try:
    score_level(105)
except ValueError as e:
    print("score_level(105) 捕获：", e)
```

任务2：用 for 和 while 各算一遍 1~100 的整数和，对比两个结果。


```python
sum_for = 0
for i in range(1, 101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print("for 求和:", sum_for)
print("while 求和:", sum_while)
print("两者结果相同:", sum_for == sum_while)
```

任务3：阶乘写两版，递归和迭代都来一个，约定 0!=1，n 取 0、1、5、10 比较。


```python
def factorial_iter(n):
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


def factorial_rec(n):
    if n <= 1:
        return 1            # 0! = 1! = 1
    return n * factorial_rec(n - 1)


for n in [0, 1, 5, 10]:
    a, b = factorial_iter(n), factorial_rec(n)
    print(f"{n:>2d}! 迭代={a:<8} 递归={b:<8} 一致={a == b}")
```

任务4：写 summary(values)，用元组把最小值、最大值、平均值一起返回，再用解包接住。


```python
def summary(values):
    return min(values), max(values), sum(values) / len(values)


low, high, average = summary([85, 58, 92, 76, 60, 89])
print("最小值:", low)
print("最大值:", high)
print("平均值:", round(average, 2))
```

任务5：写 sum_all(*args)，把参数全加起来返回，不传、传一个、传多个都试一下。


```python
def sum_all(*args):
    total = 0
    for x in args:
        total += x
    return total


print("无参数:", sum_all())
print("单参数 sum_all(5):", sum_all(5))
print("多参数 sum_all(1,2,3,4):", sum_all(1, 2, 3, 4))
```

### 实验4 文件读写、异常处理和模块

任务1：先看一下当前工作目录，拼出路径，把 exp01_data 文件夹建出来。


```python
print("当前工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录:", data_dir)
```

任务2：往 students.txt 里写一段 UTF-8 文本，再读回来，跳过表头开始解析。


```python
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
txt_path = os.path.join(data_dir, "students.txt")
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(txt_path, "r", encoding="utf-8") as f:
    raw_lines = f.read().splitlines()
print("全部行:", raw_lines)
data_lines = raw_lines[1:]          # 第一行是表头，跳过
print("跳过表头后的记录数:", len(data_lines))
```

任务3：写 parse_student(line)，检查字段个数、姓名和成绩范围，不合规就抛 ValueError。


```python
def parse_student(line):
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误（应为2）: {line!r}")
    name = parts[0].strip()
    score_text = parts[1].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_text!r}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}
```

任务4：逐行处理，把无效记录的行号和原因记下来，然后接着往下跑。


```python
valid_records = []
invalid_records = []
for line_no, line in enumerate(data_lines, start=2):   # 第 1 行是表头，所以从 2 开始数
    line = line.strip()
    if not line:
        continue
    try:
        valid_records.append(parse_student(line))
    except ValueError as err:
        invalid_records.append((line_no, str(err)))
        print(f"第 {line_no} 行无效：{err}")

print("\n有效记录：", valid_records)
print(f"有效 {len(valid_records)} 条，无效 {len(invalid_records)} 条")
```

任务5：把统计结果写进 report.txt，分别用 w 和 a 两种模式试一下。


```python
report_path = os.path.join(data_dir, "report.txt")
valid_count = len(valid_records)
invalid_count = len(invalid_records)
valid_average = sum(r["score"] for r in valid_records) / valid_count
valid_pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count
stat_line = (f"有效记录数={valid_count}, 无效记录数={invalid_count}, "
             f"平均分={valid_average:.2f}, 及格率={valid_pass_rate:.1%}\n")

# w 模式会把原来的内容清掉
with open(report_path, "w", encoding="utf-8") as f:
    f.write("【w 模式：覆盖写入，原内容被清空】\n")
    f.write(stat_line)
# a 模式是追加，原来的内容还在
with open(report_path, "a", encoding="utf-8") as f:
    f.write("【a 模式：追加写入，原内容保留】\n")
    f.write(stat_line)

print(open(report_path, encoding="utf-8").read())
```

任务6：用 pickle 把有效记录存起来再读回来，确认读回来的数据和原来的一样。

### 实验5 类、继承和对象复制

任务1：写一个 Student 类，要有 is_passed() 和 __str__() 两个方法。


```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"{self.name}: {self.score} 分（{status}）"
```

任务2：每条有效记录都建一个 Student 对象，打印出信息和是否及格。


```python
students = [Student(r["name"], r["score"]) for r in valid_records]
for stu in students:
    print(str(stu), "| is_passed() =", stu.is_passed())
```

任务3：让 GraduateStudent 继承 Student，多一个 research_topic，__str__ 重写。


```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)          # 调用父类初始化
        self.research_topic = research_topic

    def __str__(self):
        return f"研究生 {self.name}：{self.score} 分，研究课题={self.research_topic}"


grad = GraduateStudent("Alice", 85, "大数据分析")
print(grad)
print("继承自父类的 is_passed() =", grad.is_passed())
```

任务4：对一个带嵌套列表的对象，分别做直接赋值、浅拷贝和深拷贝，每次都从一个全新的原始对象开始。


```python
import copy


def copy_demo():
    original = {"name": "Alice", "scores": [85, 90]}
    alias_obj = original                # 直接赋值，还是同一个对象
    shallow = copy.copy(original)       # 浅拷贝，外面一层是新的，里面的列表还是共用的
    deep = copy.deepcopy(original)      # 深拷贝，里外都独立
    original["scores"].append(95)       # 改一下原始对象里的列表
    print("原始对象      :", original)
    print("直接赋值      :", alias_obj, " 同一外层对象:", alias_obj is original)
    print("浅拷贝 copy   :", shallow, " 内层列表共享:", shallow["scores"] is original["scores"])
    print("深拷贝 deepcopy:", deep, " 内层列表独立:", deep["scores"] is original["scores"])


copy_demo()
print("---")
copy_demo()
```
